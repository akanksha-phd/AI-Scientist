```python
# Embeddings

import torch
import torch.nn as nn

embedding = nn.Embedding(10, 4)

tokens = torch.tensor([2, 5, 7])

vectors = embedding(tokens)

print(vectors)
print(vectors.shape)
```

```python
# Attention Weights

scores = torch.tensor([2.0, 5.0, 1.0])

attention_weights = torch.softmax(scores, dim=0)

print(attention_weights)
print(attention_weights.sum())
```

```python
# Weighted Values

values = torch.tensor([
    [2.0, 1.0],
    [8.0, 6.0],
    [3.0, 4.0]
])

context = torch.sum(
    attention_weights.unsqueeze(1) * values,
    dim=0
)

print(context)
```
